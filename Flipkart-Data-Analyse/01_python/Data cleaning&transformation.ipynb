### 4.Nettoyage et normalisation des noms de colonnes

```python
data.columns = (
    data.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)
```

### 5. Nettoyage des variables catégorielles

```python
text_columns = [
    "product_name",
    "main_category",
    "sub_category",
    "seller",
    "return_policy"
]
```

```python
# supprimer les espaces inutiles
for col in text_columns:
    data[col] = data[col].astype("string").str.strip()
```

### 7.Création des variables dérivées

```python
data["estimated_revenue"] = (
    data["price"] * data["total_sold"]
)
```

```python
data["sell_through_rate"] = (
    data["total_sold"] /
    (data["total_sold"] + data["available_stock"])
)
```

```python
data["sell_through_rate"].describe()
```

### 8.Classification du stock et des réductions

```python
data["stock_status"] = np.select(
    [
        data["available_stock"] < 10,
        data["available_stock"].between(10, 50)
    ],
    [
        "Low Stock",
        "Medium Stock"
    ],
    default="High Stock"
)
```

```python
data["discount_band"] = pd.cut(
    data["discount"],
    bins=[-1, 10, 20, 30, 40, 100],
    labels=[
        "0-10%",
        "10-20%",
        "20-30%",
        "30-40%",
        "40%+"
    ]
)
```
